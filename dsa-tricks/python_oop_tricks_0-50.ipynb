```python
# 1. Class Constructio & Initialization

# 1. @dataclass for boilerplate reduction
# Use dataclass to automatically generate __init_, __repr__, and __eq__

from dataclasses import dataclass


@dataclass
class Product:
    name: str
    price: float
    in_stock: bool = True


# Sample Data
p = Product("Laptop", 999.00)
print(p)
```

```python
# 2. field(default_factory) for Mutable Defaults
# Never use mutable default arguments (like lists) directly; use default_factory

from dataclasses import dataclass, field


@dataclass
class Classroom:
    students: list = field(default_factory=list)


c = Classroom()
c.students.append("Alice")
print(c.students)
```

```python
# 3. __post_init__ for Validation Logic
# Run logic immediately after the object is initialized in a dataclass


@dataclass
class Rectangle:
    width: float
    height: float
    area: float = field(init=False)

    def __post_init__(self):
        self.area = self.width * self.height


r = Rectangle(3, 4)
print(r.area)
```

```python
# 4. @propert for Encapsulation
# Use getters/setters logic without breaking the attribute access interface


class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius

    @property
    def fahrenheit(self):
        return self._celsius * 9 / 5 + 32


t = Temperature(0)
print(t.fahrenheit)
```

```python
# 5. @property.setter for Validation
# Validate data during assignment


class Person:
    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("Age cannot be negative.")
        self._age = value


p = Person()
p.age = 30
print(p.age)
```

```python
## 2. Representation & String Formatting

# 8. __repr__ vs __str
# __repr__ should be unambiguous (for developers), __str__ readable (for users).


class User:
    def __init__(self, name):
        self.name = name

    def __repr__(self):
        return f"User('{self.name}')"

    def __str__(self):
        return self.name


u = User("Alice")
print(repr(u))  # User('Alice')
print(str(u))  # Alice
```

```python
# 9. __format__
# Hook into f-string and format()


class Currency:
    def __init__(self, amount):
        self.amount = amount

    def __format__(self, format_spec):
        if format_spec == "usd":
            return f"${self.amount:.2f}"
        return str(self.amount)


money = Currency(99.9)
print(f"{money:usd}")
```

```python
# 10. __bytes__ for Byte Representation
# Define how your object converts to bytes.


class Blob:
    def __init__(self, text):
        self.text = text

    def __bytes__(self):
        return self.text.encode("utf-8")


b = bytes(Blob("hello"))
print(b)  # b'hello'
```

```python
# 15. __iter__ for iteration


class Countdown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= 0:
            raise StopIteration
        self.current -= 1
        return self.current


for num in Countdown(3):
    print(num)
```

```python
# 16. Reversed
# Custom logic for reverse iteration


class CountDown:
    def __init__(self, start):
        self.start = start

    def __reversed__(self):
        return iter(range(self.start, -1, -1))


print(list(reversed(CountDown(3))))
```

```python
# 17. Comparison & Logic __eq__
# Compare objects by value instead of memory address


class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y


print(Point(1, 1) == Point(1, 1))
```

```python
# 18. Less than
# Define one comparison, get teh rest (<, <=, >, >=) for free

from functools import total_ordering


@total_ordering
class Account:
    def __init__(self, balance):
        self.balance = balance

    def __lt__(self, other):
        return self.balance < other.balance

    def __gt__(self, other):
        return self.balance > other.balance


a1, a2 = Account(100), Account(200)
print(a1 < a2)
print(a1 <= a2)  # Generated automatically
```

```python
# 19. __bool__ for Truthiness
# Determine if an object is True or False in boolean context


class Buffer:
    def __init__(self, items):
        self.items = items

    def __bool__(self):
        return len(self.items) > 0


b = Buffer([])
if not b:
    print("Buffer empty")
```

```python
# 20. __hash__ for Dictionary Keys
# Make objcts usable as dictionary keys (must be immutable)


class Coordinate:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __eq__(self, other):
        return (self.x, self.y) == (other.x, other.y)

    def __hash__(self):
        return hash((self.x, self.y))


d = {Coordinate(1, 1): "Home"}
print(d[Coordinate(1, 1)])
```

```python
# 22. __getattribute__ for Total Control
# Called every time an attribute is accessed (careful with recursion).


class Logger:
    def __getattribute__(self, name):
        print(f"Accessing {name}")
        return super().__getattribute__(name)

    def value(self):
        return 42

    def greeting(self):
        return "hello"


l = Logger()
print(l.value())
print(l.greeting())
```

```python
# 23. __setattr__ for Intercepting Assignment
# Called on every attribute assignment


class Protected:
    def __setattr__(self, name, value):
        if name == "id" and hasattr(self, "id"):
            raise AttributeError("ID cannot be changed")
        super().__setattr__(name, value)


p = Protected()
p.id = 1
p.id = 2  # Raise error
```

```python
# 28. Inheritance & Structure
# Ensure all parent classes are initialized correctly in diamond inheritance.


class A:
    def __init__(self):
        print("A")


class B(A):
    def __init__(self):
        print("B")
        super().__init__()


class C(A):
    def __init__(self):
        print("C")
        super().__init__()


class D(B, C):  # print B first, then C
    def __init__(self):
        print("D")
        super().__init__()


d = D()
```

```python
# 31. Protocol (Structural Subtyping)
# Duck typing with type hints. If it walks like a duck, it is a duck.

from typing import Protocol


class Flyer(Protocol):
    def fly(self) -> None: ...


class Bird:
    def fly(self):
        print("Flying")


def make_it_fly(obj: Flyer):
    obj.fly()


make_it_fly(Bird())  # Works without explicit inheritance
```

```python
# 33. @classmethod for Alternative Constructors
# Factory methods that returninstances of the class

import time


class Date:
    def __init__(self, year, month, day):
        self.year, self.month, self.day = year, month, day

    @classmethod
    def today(cls):
        t = time.localtime()
        return cls(t.tm_year, t.tm_mon, t.tm_mday)


d = Date.today()
print(d.year)
```

```python
# 37. Lazy Import/Loading
# Import a module or create an object only when accessed.


class LazyLoader:
    def __init__(self, module_name):
        self.module_name = module_name
        self._module = None

    def __getattr__(self, name):
        if self._module is None:
            import importlib

            self._module = importlib.import_module(self.module_name)
        return getattr(self._module, name)


json = LazyLoader("json")
# json module is only actually imported when we call json.dumps
```

```python
# 39. functools.singledispatchmethod
# Method overloading based on the type of the first argument

from functools import singledispatchmethod


class Formatter:
    @singledispatchmethod
    def format(self, arg):
        raise NotImplementedError("Unsupported type")

    @format.register
    def _(self, arg: int):
        return f"Integer: {arg}"

    @format.register
    def _(self, arg: str):
        return f"String: {arg!r}"


f = Formatter()
print(f.format(10))
print(f.format("hi"))
```

```python
# 40. __copy__ and __deepcopy__
# Customize how copy.copy and copy.deepcopy behave for your objects.

import copy


class Node:
    def __init__(self, value):
        self.value = value

    def __copy__(self):
        print("Shallow copy")
        return Node(self.value)

    def __deepcopy__(self, memo):
        print("Deep copy")
        return Node(copy.deepcopy(self.value, memo))


n = Node(5)
copy.copy(n)
copy.deepcopy(n)
```

```python
# 42. __radd__ for Right-Side Addition
# Handle cases where your object is on the right side of the operator (10 + obj)


class Money:
    def __init__(self, amount):
        self.amount = amount

    def __radd__(self, other):
        return Money(self.amount + other)


m = Money(10)
total = 10 + m  # Calls m.__radd__(10)
print(total.amount)
```

```python
# 50. __reduce__ for Pickling
# Customize how an object is serialized by pickle.

import pickle


class Thing:
    def __init__(self, data):
        self.data = data

    def __reduce__(self):
        # Return (callable, args) to recreate object
        return (self.__class__, (self.data,))


t = Thing("save_me")
blob = pickle.dumps(t)
new_t = pickle.loads(blob)
print(new_t.data)
```
