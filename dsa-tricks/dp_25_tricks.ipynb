```python
# 1. Fibonacci dp
def fib(n):
    dp = [0]*(n+1)
    dp[1] = 1
    for i in range(2, n+1):
        dp[i] = dp[i-1] + dp[i-2]
    return dp[n]

def fib_optimized(n):
    a, b = 0, 1   # use variables only
    for i in range(2, n+1):
        a, b = b, a+b
    return b
    
n = 10
print(fib(n))
print(fib_optimized(n))
```

```python
# 2. O/1 Knapsack
def knapsack(w, val, W):
    n = len(w)
    dp = [[0]*(W+1) for _ in range(n+1)]
    for i in range(1, n+1):
        for j in range(W+1):
            dp[i][j] = dp[i-1][j]
            if j >= w[i-1]:
                dp[i][j] = max(dp[i][j], dp[i-1][j-w[i-1]] + val[i-1])
    return dp[n][W]

weights = [2,3,4]
values = [4,5,6]
capacity = 6

print(knapsack(weights, values, capacity))
```

```python
# 3. Unbounded Knapsack
# Input
weights=[2,3,4]
values=[4,5,6]
capacity=6

dp = [0]*(capacity+1)

for i in range(len(weights)):
    for j in range(weights[i], capacity+1):
        dp[j] = max(dp[j], dp[j-weights[i]] + values[i])

print(dp[capacity])
```

```python
# 4. Coin Change (ways)
# How many ways were there to make the amount
# Input
coins = [1,2,5]
amount = 5

# Code
dp = [0] * (amount + 1)
dp[0] = 1
for coin in coins:
    for i in range(coin, amount+1):
        dp[i] += dp[i-coin] # reuseand and add to current total
        print(dp)

print(dp[amount], dp)
```

```python
# 5. Coin Change (Minimum Coins)
# Input
coins = [1,2,5]
amount = 5

# Code
INF = 10**9
dp = [INF] * (amount + 1)
dp[0] = 0

for coin in coins:
    for i in range(coin, amount+1):
        dp[i] = min(dp[i], dp[i-coin]+1)

print(dp[amount])
```

```python
# 6. Longest Increasing Subsequence
arr = [3,1,4,1,5]

# Code
n = len(arr)
dp = [1] * n

for i in range(n):  # loop through each element, find LIS at that specific position
    for j in range(i):  # backward before current position
        if arr[j] < arr[i]:  # conditionn for increasing subsequence, only consider this
            dp[i] = max(dp[i], dp[j] + 1)  # update step, store the best possible (longest)

print(max(dp))
```

```python
# 7. LIS (Binary Search)
# transitions from the $O(n^2)$ 
# Dynamic Programming approach to a much faster $O(n \log n)$ 
# method using Binary Search

arr = [3,1,4,1,5]

# Code
import bisect
lis = []  # store the "best" (smallest) ending elements for increasing subsequences found so far.

for x in arr:
    # The Binary Search: This finds the index 
    # where x should be inserted into lis to keep 
    # it sorted. It's much faster than checking 
    # every previous element.
    pos = bisect.bisect_left(lis, x)
    # If pos is equal to the current length of lis, 
    # it means x is larger than 
    # any element we've seen so far.
    if pos == len(lis):
        lis.append(x) # Since x is the largest, we extend our longest subsequence by adding x to the end.
    else:
        lis[pos] = x

# execution
print(len(lis))
        
```

```python
# 8. Longest Common Subsequence
# Input
a = "abcde"
b = "ace"

# Code
n = len(a)
m = len(b)

dp = [[0] * (m+1) for _ in range(n + 1)]

for i in range(1, n+1):
    for j in range(1, m+1):
        if a[i-1] == b[j-1]:
            dp[i][j] = dp[i-1][j-1]+1
        else:
            dp[i][j] = max(dp[i-1][j], dp[i][j-1])

# Execution 
print(dp[n][m])
```

```python
# 9. Edit Distance
# Input
a = "abcde"
b = "ace"

# Code
n = len(a)
m = len(b)

dp = [[0] * (m+1) for _ in range(n+1)]

for i in range(n+1):
    dp[i][0] = i

for j in range(m+1):
    dp[0][j] = j

for i in range(1, n+1):
    for j in range(1, m+1):
        if a[i-1] == b[j-1]:
            dp[i][j] = dp[i-1][j-1]
        else:
            dp[i][j] = 1 + min(
                dp[i-1][j],
                dp[i][j-1],
                dp[i-1][j-1]
            )

print(dp[n][m])
```

```python
# 10 Subset Sum
arr = [3,1,4,1,5]
target = 7

# Code
dp = [False] * (target + 1)
dp[0] = True

for x in arr:
    for j in range(target, x-1, -1):
        dp[j] = dp[j-x]

# Execution 
print(dp[target])
```

```python
# 11 Maximum Subarray (Kadane)
# find the Maximum Subarray Sum in $O(n)$ time.
# Input 
arr = [-2,1,-3,4,-1,2,1,-5,4]

# Code
best = cur = arr[0]

for x in arr[1:]:   # starting from second element
    # local decision, is it better to start a brand new subarray
    # or add this numbr to existing subarray built
    # if cur+x < x (x is negative), reset and start over from x
    cur = max(x, cur+x)
    # global tracker, best sum ending right here
    # best remembers the peak ever seen through the entire process
    best = max(best, cur)

print(best)
```

```python
# 11. Unique Grid Paths (only move Right or Down)
# Input 
n = 3
m = 3

# Code
# Only 1 way to stay at very top edge / very left edge
dp = [[1]*m for _ in range(n)]

# start at first row (i=0) and first col (j=0)
# From above cell (i-1, j)
# From left cell (i, j-1)
# total = sum of the ways to reach those two neighbors.
for i in range(1, n):
    for j in range(1, m):
        print(i, j, dp)
        dp[i][j] = dp[i-1][j] + dp[i][j-1]
        print(i, j, dp)

# Executinon
print(dp[n-1][m-1])
```

```python
# 13. Grid Maximum Path Sum
# Input
grid = [
    [1,3,1],
    [1,5,1],
    [4,2,1]
]

# Code
n = len(grid)
m = len(grid[0])

dp = [[0]*m for _ in range(n)]

for i in range(n):
    for j in range(m):
        dp[i][j] = grid[i][j]

        if i > 0:
            dp[i][j] = max(dp[i][j], grid[i][j]+dp[i-1][j])

        if j > 0:
            dp[i][j] = max(dp[i][j], grid[i][j]+dp[i][j-1])

# Execution 
print(dp[n-1][m-1])
```

```python
# 15. Traveling Salesman (DP)
# Input
dist=[
[0,10,15,20],
[10,0,35,25],
[15,35,0,30],
[20,25,30,0]
]

# Code
n = len(dist)
INF = 10**9

dp = [[INF]*n for _ in range(1<<n)]
dp[1][0] = 0

for mask in range(1<<n):
    for u in range(n):
        if mask & (1<<u):
            for v in range(n):
                if not mask & (1<<v):
                    dp[mask|1<<v][v] = min(
                        dp[mask|1<<v][v],
                        dp[mask][u]+dist[u][v]
                    )
res = min(dp[(1<<n)-1][i] + dist[i][0] for i in range(n))
print(res)
```

```python
# 16 Digit DP Example

# Input
n = 100

# Code
count = 0

for i in range(n+1):
    if "4" not in str(i):
        count += 1
print(count)

# 16. Digit DP Solution (Counting numbers without the digit '4')
n_str = str(100)
memo = {}

def count_valid(index, is_less, is_started):
    # Base case: reached the end of the number
    if index == len(n_str):
        return 1
    
    state = (index, is_less, is_started)
    if state in memo:
        return memo[state]
    
    res = 0
    # Determine the upper limit for the current digit
    limit = int(n_str[index]) if not is_less else 9
    
    for digit in range(limit + 1):
        # The core condition: skip the digit '4'
        if digit == 4:
            continue
            
        new_is_less = is_less or (digit < limit)
        new_is_started = is_started or (digit > 0)
        
        res += count_valid(index + 1, new_is_less, new_is_started)
        
    memo[state] = res
    return res

print(count_valid(0, False, False))
```

```python
# Palindrome DP
# Input
s = "babad"

# Code
n = len(s)
dp = [[False]*n for _ in range(n)]

start = 0 
max_len = 1

for i in range(n):
    dp[i][i] = True

for l in range(2, n+1):
    for i in range(n - l+1):
        j = i + l - 1
        if s[i] == s[j] and (l == 2 or dp[i+1][j-1]):
            dp[i][j] = True

            if l > max_len:
                start = i 
                max_len = l

# Execution 
print(s[start:start + max_len])
```
