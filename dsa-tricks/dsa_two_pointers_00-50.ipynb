```python
# 7. Valid palindrome (opposite direction, converging pointers)
# Standard technique for sorted arrays or palindromes where pointers start at ends and move inward.
# Checks if a string reads the same forwards and backwards, ignoring non-alphanumeric characters.

def is_palindrome(s: str) -> bool:
    l, r = 0, len(s) -1
    while l < r:
        while l < r and not s[l].isalnum():
            l += 1
        while l < r and not s[r].isalnum():
            r -= 1
        if s[l].lower() != s[r].lower():
            return False
        l, r = l + 1, r - 1
    return True

print(is_palindrome('aabaa'))
print(is_palindrome('hello'))
        
```

```python
# 8. Reverse string
# Reverses a string in-place using tuple unpacking.
# To reverse a string using your two-pointer logic, 
# you must first convert the string into a list (which is mutable), 
# perform the swaps, and then join it back into a string.

def reverse_string(s: list) -> None:
    # Convert string to a list to allow item assignment
    chars = list(s)
    l, r = 0, len(s) -1
    while l < r:
        chars[l], chars[r] = chars[r], chars[l]
        l, r = l+1, r-1
    # join the list back to string
    return ''.join(chars)

s = 'abc'
print(reverse_string(s))
s = 'xyz'
print(reverse_string(s))
```

```python
# 15. Sort colors (Dutch National Flag)
# Sorts array of 0s, 1s, and 2s in one pass
def sort_colors(nums: list[int]) -> None:
    low, mid, high = 0, 0, len(nums) -1
    while mid <= high:
        if nums[mid] == 0:
            nums[low], nums[mid] = nums[mid], nums[low]
            low += 1
            mid += 1
        elif nums[mid] == 1:
            mid += 1
        else:
            nums[mid], nums[high] = nums[high], nums[mid]
            high -= 1
    return nums

print(f"Sorted: {sort_colors([2, 0, 2, 1, 1, 0])}")
```

```python
# 16. Valid palindrome II 
# Checks if palindrome is possible by deleting at most one character.

def valid_palindraome_ii(s: str) -> bool:
    def check(l, r):
        while l < r:
            if s[l] != s[r]: return False
            l, r = l+1, r+1
        return True

    l, r = 0, len(s) -1
    while l < r:
        if s[l] != s[r]:
            return check(l+1, r) or check(l, r-1)
        l, r = l+1, r-1
    return True
```

```python
# 17. 3Sum closest
# Find triplet with sum closest to target

def three_sum_closest(nums: list[int], target: int) -> int:
    nums.sort()
    closest = float('inf')
    for i in range(len(nums) - 2):
        l, r = i +1, len(nums) -1
        while l < r:
            current = nums[i] + nums[l] + nums[r]
            if abs(target - current) < abs(target - closest):
                closest = current
            if current < target: l += 1
            elif current > target: r -= 1
            else: return current
        return closest

nums = [1,2,3,9,4,5,6,7,8]
target = 18
three_sum_closest(nums, target)
```

```python
# 18. 4Sum
# Find all unique quadruplest summing to target

def four_sum(nums: list[int], target: int) -> list[list[int]]:
    nums.sort()
    n, res = len(nums), []
    for i in range(n-3):
        if i > 0 and nums[i] == nums[i-1]: continue
        for j in range(i+1, n-2):
            if j > i + 1 and nums[j] == nums[j-1]: continue
            l, r = j +1, n - 1
            while l < r:
                s = nums[i] + nums[j] + nums[l] + nums[r]
                if s == target:
                    res.append([nums[i], nums[j], nums[l], nums[r]])
                    while l < r and nums[l] == nums[l+1]: l +=1
                    l, r = l+1, r -1
                elif s < target: l += 1
                else: r -= 1
        return res 

nums = [1,2,3,9,4,5,6,7,8]
target = 22
four_sum(nums, target)
```

```python
# 19. Reverse words in a string II
# Reverses words in-place (assuming list of chars).

def reverse_words(words: list[str]) -> None:
    s = list(words)
    def reverse_range(l, r):
        while l < r:
            s[l], s[r] = s[r], s[l]
            l, r = l + 1, r - 1
    reverse_range(0, len(s) - 1)  # reverse whole string
    start = 0 
    for i in range(len(s) + 1):
        if i == len(s) or s[i] == ' ':
            reverse_range(start, i - 1)  # reverse each word
            start = i + 1
    return  s
    
words = 'hello world'
print(reverse_words(words))
```

```python
# 20. Longest Palindromic Substring
# Expands around center (variation of two pointers)

def longest_palindrome(s: str) -> str:
    res = ""
    for i in range(len(s)):
        # Odd length
        l, r = i, i
        while l >= 0 and r <len(s) and s[l] == s[r]:
            if (r - l + 1) > len(res): res = s[l:r+1]
            l, r = l-1, r+1   # expand outwards
        # Even length
        l, r = i, i + 1
        while l >= 0 and r < len(s) and s[l] == s[r]:
            if (r - l + 1) > len(res): res = s[l:r+1]
            l, r = l-1, r+1
    return res 

s = 'abacabbad'
longest_palindrome(s)
```

```python
# 21. Minimum Difference Between Two Elements
# Find min differences between a sorted array

def min_max_diff_sorted(arr: list[int]) -> tuple:
    arr.sort()
    min_diff = float('inf')
    max_diff = 0
    for i in range(1, len(arr)):   # start from 1
        min_diff = min(min_diff, arr[i] - arr[i-1])
        max_diff = max(max_diff, arr[i] - arr[i-1])
    return min_diff, max_diff

arr = [1,2,3,6,9]
print(min_max_diff_sorted(arr))
```

```python
# 23. Partition Array (Quick Sort Logic)
# Partition array based on pivot

def partition(nums: list[int], pivot_idx: int) -> (int, list):
    pivot = nums[pivot_idx]
    nums[pivot_idx], nums[-1] = nums[-1], nums[pivot_idx]
    store_idx = 0 
    for i in range(len(nums) -1):
        if nums[i] < pivot:
            nums[store_idx], nums[i] = nums[i], nums[store_idx]
            store_idx += 1
    nums[store_idx], nums[-1] = nums[-1], nums[store_idx]
    return store_idx, nums

arr = [1,2,9,3,6]
pivot_idx = 4
partition(arr, pivot_idx)
```

```python
# 24. Next Permutation
# Rotates array to the right by k steps

"""
The Best Fix: The Three-Reversal Trick

The most efficient way to 
rotate in-place ($O(1)$ space, $O(n)$ time) is to use three 
specific reversals.The Logic:
Reverse the entire array.
Reverse the first $k$ elements.
Reverse the remaining $n-k$ elements.

The Trick: The reversal happens in-place. 
You only ever use a single temporary variable to 
swap two numbers at a time, regardless of how large the array is.

"""
def rotate(nums: list[int], k: int) -> list:
    n = len(nums)
    # 1. Handle cases where k > n
    k %= n  

    # Helper to reverse a portion of the array in-place
    def reverse(l, r):
        while l < r: 
            nums[l], nums[r] = nums[r], nums[l]
            l, r = l+1, r-1   # move inwards
    # 2. Apply the three reversals
    reverse(0, n-1)    # Reverse everything
    reverse(0, k-1)    # Reverse first k elements
    reverse(k, n-1)    # Reverse the rest
    return nums

arr = list(range(10))
k = 5
print(rotate(arr, k))
```

```python
# 30. Happy Number
# Determine if a number is "happy"

def is_happy(n: int) -> bool:
    def get_next(n):
        return sum(int(d)**2 for d in str(n))
    slow, fast = n, get_next(n)
    while fast != 1 and slow != fast:
        slow = get_next(slow)
        fast = get_next(get_next(fast))
    return fast == 1

print(is_happy(19))
print(is_happy(2))

"""
Sample Trace: $n = 19
$$1^2 + 9^2 = 1 + 81 = 82
$$8^2 + 2^2 = 64 + 4 = 68
$$6^2 + 8^2 = 36 + 64 = 100
$$1^2 + 0^2 + 0^2 = 1$ $\rightarrow$ Happy!

Sample Trace: $n = 2$ (Unhappy)
$ 2^2 = 4
$$4^2 = 16
$$1^2 + 6^2 = 37
$... 
eventually hits 4 again, 
creating a loop: 4, 16, 37, 58, 89, 145, 42, 20, 4.
"""

# Happy Number Using HashSet Approach (O(n)) space
def is_happy(n: int) -> bool:
    seen = set()
    while n != 1 and n not in seen:
        seen.add(n)   # Add no not a happy number
        # Calculate sum fo squares of digits
        n = sum(int(digit)**2 for digit in str(n))
    return n == 1

print(is_happy(19))
print(is_happy(2))
```

```python
# 31. Find the Duplicate Number
# Finds duplicate in array containing numbers 1 to n.

def find_duplicate(nums: list[int]) -> int:
    # Phase 1: Finding the intersection point of the cycle    
    slow, fast = nums[0], nums[0]
    while True:
        slow = nums[slow]
        fast = nums[nums[fast]]
        if slow == fast: break
    # Phase 2: Finding the "entrance" to the cycle (the duplicate)            
    slow = nums[0]
    while slow != fast:
        slow = nums[slow]
        fast = nums[fast]
    return slow


# Standard Case
print(find_duplicate([1, 3, 4, 2, 2]))  # Expected: 2

# Duplicate at the beginning/end
print(find_duplicate([3, 1, 3, 4, 2]))  # Expected: 3

# Multiple repetitions of the same number
print(find_duplicate([1, 1, 1, 1, 1]))  # Expected: 1
```

```python
# Category 3: Sliding Window (Same Direction)
# Used for contiguous subarray problems

# 34. Minimum Size Subarray Sum
# Find minimal length subarray with sum >= target

def min_sub_array_len(target: int, nums: list[int]) -> int:
    l, current_sum, min_len = 0, 0, float('inf')
    for r in range(len(nums)):
        current_sum += nums[r]
        while current_sum >= target:
            min_len = min(min_len, r - l + 1)
            current_sum -= nums[l]
            l += 1
    return min_len if min_len != float('inf') else 0
    
```

```python
# 35. Longest Substring Without Repeating Characters

def length_of_longest_substring(s: str) -> int:
    char_index = {}
    l, max_len = 0, 0
    for r, char in enumerate(s):
        if char in char_index and char_index[char] >= l:
            l = char_index[char] + 1
        char_index[char] = r
        max_len = max(max_len, r - l + 1)
    return max_len
```

```python
# 36. Permutation in String
# Checks if s3 contains permutation of s1

def check_inclusion(s1: str, s2: str) -> bool:
    cntr, w, matched = collections.Counter(s1), len(s1), 0
    for i in range(len(s2)):
        if s2[i] in cntr:
            cntr[s2[i]] -= 1
            if cntr[s2[i]] == 0: matched += 1
        if i >= w and s2[i-w] in cntr:
            if cntr[s2[i-w]] == 0: matched -= 1
            cntr[s2[i-w]] += 1
        if matched == len(cntr): return True
    return False

    
```

```python
# 37. Find All Anagrams in a String
def find_anagram(s: str, p: str) -> list[int]:
    res, p_count = [], collections.Counter(p)
    s_count = collections.Counter(s[:len(p)])
    if s_count == p_count: res.append(0)
    for i in range(len(p), len(s)):
        s_count[s[i]] = s_count.gt(s[i], 0) + 1
        s_count[s[i-len(p)]] == 1
        if s_count[s[i-len(p)]] == 0: del s_count[s[i-len(p)]]
        if s_count == p_count: res.append(i - len(p) + 1)
    return res
    
```

```python
# 38. Max Consecutive Ones III
# Longest subarray with at most K zeros

def longest_ones(nums: list[int], k: int) -> int:
    l, zeros, max_len = 0, 0, 0
    for r in range(len(nums)):
        if nums[r] == 0: zeros += 1
        while zeros > k:
            if nums[l] == 0: zeros -= 1
            l += 1
        max_len = max(max_len, r - l + 1)
    return max_len
```

```python
# 39. Longest Repeating Character Replacement

def character_replacement(s: str, k: int) -> int:
    count, l, res, max_f = {}, 0, 0, 0
    for r in range(len(s)):
        count[s[r]] = count.get(s[r], 0) + 1
        max_f = max(max_f, count[s[r]])
        while (r - l + 1) - max_f > k:
            count[s[l]] -= 1
            l += 1
        res = max(res, r - l + 1)
    return res
    
```

```python
# 40. Subarray Product Less Thank K
def num_subarray_product_less_than_k(nums: list[int], k: int) -> int:
    if k <= 1: return 0
    prod, l, res = 1, 0, 0
    for r in range(len(nums)):
        prod *= nums[r]
        while prod >= k:
            prod //= nums[l]
            l += 1
        res += r - l + 1
    return res
    
    
```

```python
# 42. Grumpy Bookstore Owner
def max_satisfied(customers: list[int], grumpy: list[int], minutes: int) -> int:
    n = len(customers)
    # Base satisfied
    satisfied = sum(c for i, c in enumerate(customers) if grumpy[i] == 0)

    # Sliding window for extra satisfied
    extra = 0 
    for i in range(minutes):
        if gumpy[i] == 1: extra += customers[i]
    max_extra = extra

    for i in range(minutes, n):
        if grumpy[i] == 1: extra += customers[i]
        if grumpy[i-minutes] == 1: extra -= customers[i-minutes]
        max_extra = max(max_extra, extra)
    return satisfied + max_extra
```

```python
# Category 4: Array Modification / Partitioning
# Two pointers moving same direction to filter or modify arrays

# 44. Remove Element
# Removes occurrences of val in-place
def remove_element(nums: list[int], val: int) -> int:
    k = 0
    for i in range(len(nums)):
        if nums[i] != val:
            nums[k] = nums[i]
            k += 1
    return k
```

```python
# 45. Remove Duplicates from Sorted Array

def remove_duplicates(nums: list[int]) -> int:
    if not nums: return 0
    write = 1
    for read in range(1, len(nums)):
        if nums[read] != nums[read-1]:
            nums[write] = nums[read]
            write +=1
    return write
            
```

```python
# 46. Remove Duplicates from Sorted Array II (Allow duplicates twice)

def remove_duplicates_ii(nums: list[int]) -> int:
    if len(nums) <= 2: return len(nums)
    write = 2
    for read in range(2, len(nums)):
        if nums[read] != nums[write-2]:
            nums[write] = nums[read]
            write += 1
    return write
    
    
```

```python
# 47. Apply Operations to an Array
# Mutiply equal adjacent numbers and move zeros

def apply_operations(nums: list[int]) -> list[int]:
    for i in range(len(nums) -1):
        if nums[i] == nums[i+1]:
            nums[i] *= 2
            nums[i+1] = 0
    # Move zeros
    pos = 0
    for x in nums:
        if x != 0:
            nums[pos] = x
            pos += 1
    for i in range(pos, len(nums)):
        nums[i] = 0
    return nums
    
```

```python
# 48. Sort Array By Parity
# Evens first, odds last

def sort_array_by_parity(nums: list[int]) -> list[int]:
    l, r = 0, len(nums) -1
    while l < r:
        while l < r and nums[l] % 2 == 0: l += 1
        while l < r and nums[l] % 2 == 1: r -= 1
        nums[l], nums[r] = nums[r], nums[l]
        l, r = l + 1, r + 1
    return nums
```

```python
# 49. Sort Array By Parity II
def sort_array_by_parity_ii(nums: list[int]) -> list[int]:
    even, odd = 0, 1
    res = [0] * len(nums)
    for x in nums:
        if x % 2 == 0:
            res[even] = x
            even += 2
        else:
            res[odd] = x
            odd += 2
    return res
    
```

```python
# Category 5: Two Arrays / Merge
# Pointers iterating over two different data structures

# 50. Merged Sorted Array
# Merges nums2 into nums1

def merge(nums1: list[int], m: int, nums2: list[int], n: int) -> None:
    p1, p2, write = m - 1, n - 1, m + n - 1
    while p2 >= 0:
        if p1 >= 0 and nums1[p1] > nums2[p2]:
            nums1[write] = nums[p1]
            p1 -= 1
        else:
            nums1[write] = nums2[p2]
            p2 -= 1
        write -= 1
    
```

```python
# 51. Intersection of Two Arrays

def intersection(nums1: list[int], nums2: list[int]) -> list[int]:
    nums1.sort(); nums2.sort()
    i, j, res = 0, 0, set()
    while i < len(nums1) and j < len(nums2):
        if nums1[i] < nums2[j]: i += 1
        elif nums1[i] > nums2[j]: j += 1
        else: 
            res.add(nums1[i])
            i += 1
            j += 1
    return list(res)
```

```python
# 52. Interval List Intersections

def interval_intersection(firstList: list[list[int]], secondList: list[list[int]]) -> list[list[int]]:
    i, j, res = 0, 0, []
    while i < len(firstList) and j < len(secondList):
        start = max(firstList[i][0], secondList[j][0])
        end = min(firstList[i][1], secondList[j][1])
        if start <= end: res.append([start, end])
        if firstList[i][1] < secondList[j][1]: i += 1
        else: j += 1
    return res
```
