```python
# 151. Memorization with lru_cache

from functools import lru_cache

@lru_cache
def fib(n):
    if n < 2: 
        return n
    return fib(n-1) + fib(n-2)

fib(50)
```

```python
# 152. Custom sorting with functools.cmp_to_key
# Use a comparator function (returns -1, 0, 1) for complex sorting logic.

from functools import cmp_to_key

def compare(a, b):
    if a % 2 != b % 2: 
        return b % 2 - a % 2  # odds first
    return a - b

arr = [4,1,3,2,7,8,0,5]
arr.sort(key=cmp_to_key(compare))
print(arr) 

# The Breakdown:Evens: 2, 4 (sorted ascending $\rightarrow$ 2, 4)
# Odds: 1, 3 (sorted ascending $\rightarrow$ 1, 3)
# Combined (Evens then Odds): [2, 4, 1, 3]
```

```python
# 153. divmod (a//b, a%b) for Grid Traversal

#Calculate row and column from a 1D index in one step.

index, cols = 15, 4
row, col = divmod(index, cols)
print(f"{row}, {col}")

# Why this is useful for Grid TraversalIf you have a grid with a fixed number of columns (in this case, 4), 
# and you are treating it as one long continuous list, divmod tells you exactly where any index sits in 2D 
# space:
# Index  0  1  2  3
# Row 0  0  1  2  3
# Row 1  4  5  6  7
# Row 2  8  9  10 11
# Row 3  12 13 14 15
```

```python
# 179. yield from for Recursive Generators (clean recursive iteration)
def flatten(arr):
    for x in arr:
        if isinstance(x, list):
            yield from flatten(x)
        else:
            yield x

print(list(flatten([1, [2, [3,4]]])))        
```

```python
# 183. Isolate lowest set bit (x & -x, Useful for Fenwick Trees)
x = 12 # 1100
lsb = x & -x
print(lsb, bin(x), bin(x & -x))

'''
The Logic: x & -xIn your code, x = 12. 
Let's look at what happens at the binary level:
Binary of x (12): 
0000 1100 (The lowest set bit is at the $2^2$ position, which is 4).
Binary of -x (-12):
Python uses Two's Complement to represent negative numbers. 
To get -12, you invert the bits of 12 and add 1:
Invert 1100 $\rightarrow$ ...
1111 0011
Add 1 $\rightarrow$ ...
1111 0100
The Bitwise AND (&):Now, compare x and -x bit by bit:Plaintext  
0000 1100  (12)
& 
1111 0100  (-12)
-----------
0000 0100  (Result = 4)
'''
```

```python
# 184. Check power of two (Classic bit trick)
def is_power_of_two(n):
    print(n, bin(n), bin(n-1))
    return n > 0 and (n & (n - 1) == 0)
print(is_power_of_two(16))
```

```python
# 185. Bitmasks with << and | (Set specific bits)
def set_bit(num, i):
    print(num, (1<<i), bin(1<<i))
    return num | (1 << i)
print(set_bit(0, 2))
```
