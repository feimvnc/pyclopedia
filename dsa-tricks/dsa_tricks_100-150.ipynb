```python
# 103. yield from for Recursive Generators

# When writing recursive traversals (like DFS on a tree), 
# building a list and returning it is memory-inefficient. 
# Using yield creates a generator. The yield from syntax allows 
# you to delegate iteration to a sub-generator cleanly.

class Node: 
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def traverse(root):
    if root:
        yield root.val
        # Recursively yield everything from left, then right
        yield from traverse(root.left)
        yield from traverse(root.right)

# Usage
root = Node(1, Node(2), Node(3))
print(list(traverse(root)))
```

```python
# 111. match / case for State Machines (python 3.10+)

# Python 3.10 introduced Structural Pattern Matching. This is vastly 
# superior to long if-elif-else chains for parsing instructions, 
# handling graph node states, or decoding inputs.

def handle_command(command):
    match command:
        case "quit" | "exit":
            print("Goodbye!")
        case ("move", direction) if direction in "NSEW":
            print(f"Moving {direction}")
        case ("attack", target):
            print(f"Attacking {target}")
        case _:
            print("Unknown command")

handle_command(("move", "N"))

```

```python
# 124. Bit Masking (Flags)
# Permissions
READ = 1
WRITE = 2
EXECUTE = 3 
# Grand Read and Write permissions
user_perms = READ | WRITE
print(f"{user_perms=}, {bin(user_perms)}")

# Check if user has Read permission 
if user_perms & READ:
    print("User can read.")

# Check if user has execute permission
if user_perms & EXECUTE:
    print("User can execute")

# Toggle Write permission (Remove it if they have it, add if not)
user_perms = user_perms ^ WRITE
print(f"After toggle: {user_perms}")

'''
Summary Table for Quick Reference
Goal              Code Snippet
Multiply by 2	  x << 1
Divide by 2	      x >> 1
Check if Odd	  (x & 1) == 1
Get i -th bit	  (x >> i) & 1
Set i -th bit	  `x
Clear i -th bit	  x & ~(1 << i)
Clear lowest set bit	x & (x - 1)
Extract lowest set bit	x & -x
'''
```

```python
# 129. One-liner memorization (functools.lru_cache)

# Dynamic Programming often requires a 
# memoization dictionary. Python has a built-in decorator 
# that handles the cache automatically.

from functools import lru_cache 

@lru_cache(maxsize=None) # Unlimited cache
def fib(n):
    if n < 2: 
        return n
    return fib(n-1) + fib(n-2)

print(fib(10))
```

```python
# 143. functools.cmp_to_key for complex sorting

# Sometimes a simple key=lambda isn't enough (e.g.,  
# problems asking for "custom sort order"). cmp_to_key converts an old-style 
# comparison function (returning -1, 0, 1) into a key function.

from functools import cmp_to_key

# Custom comparator: Sort by absolute value
def compare(a, b):
    if abs(a) < abs(b): return -1
    if abs(a) > abs(b): return 1
    return 0

arr = [-5, 2, -1, 8]
arr.sort(key=cmp_to_key(compare))
print(arr)
```

```python
# 150. 10. EAFP (Easier to Ask for Forgiveness than Permission)

# In Python, it's often idiomatic to try an operation and catch an exception 
# rather than checking pre-conditions (Look Before You Leap). 
# This is particularly useful when accessing dictionary keys or handling parsing.

data = {'a': 1, 'b': 2}

# LBYL (Look Before You Leap) - Verbose
if 'c' in data:
    x = data['c']
else:
    x = 0

# EAFP - Idiomatic Python
try:
    x = data['c']
except KeyError:
    x = 0
    
# Note: For simple defaults, data.get('c', 0) is best, 
# but EAFP is useful when the operation is a complex logic block 
# or the check is as expensive as the operation itself.

```
