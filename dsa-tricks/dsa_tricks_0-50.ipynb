```python
# 1. Use collections.Counter instead of manual counting dictionaries

# Instead of 
lst = [3,1,4,1,5]
freq = {}
for x in lst:
    freq[x] = freq.get(x, 0) + 1
print(freq)

# Do:
from collections import Counter
freq = Counter(lst)  # most common idiom
print(freq)         
print(freq.most_common(3))
```

```python
# 2. defaultdict is your friend for grouping

from collections import defaultdict

# Instead of 
pairs = [(3,1), (4,1), (5,9)]
groups = {}
for k, v in pairs:
    if k not in groups:
        groups[k] = []
    groups[k].append(v)
print(groups)

# Do:
groups = defaultdict(list)
for k, v in pairs:
    groups[k].append(v)
print(groups)
```

```python
# 6. Using any() + generator expression instead of for + break
# Instead of:
items = [3,1,4,1,5]

def expensive_check(x):
    return x >= 5
    
found = False
for x in items:
    if expensive_check(x):
        found = True
        print("Expensive item")
        break

# Do:
if any(expensive_check(x) for x in items):
    print("Expensive item")
```

```python
# 19. deque for O(1) sliding-window-maximum style (common questions)
from collections import deque

nums = [1, 3, -1, -3, 5, 3, 6, 7]
k = 3 

dq = deque()  # will store indices
result = []

for i, num in enumerate(nums):
    # remove elememts smaller than current from back
    while dq and nums[dq[-1]] < num:
        dq.pop()
    dq.append(i)

    # remove out-of-window, dp[0] must > i - k when adding new item
    if dq[0] == i - k:
        dq.popleft() 

    # after window is formed
    if i >= k - 1:
        result.append(nums[dq[0]])   # front is always the maximum

print(result)
```

```python
# 52. functools.cached_property (Python 3.8+)
# In algorithms involving objects (like nodes), you might compute a value 
# (e.g., is_leaf) frequently. @cached_property calculates it once and stores 
# it as an attribute, avoiding recomputation.

from functools import cached_property

class Node:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left 
        self.right = right

    @cached_property
    def is_leaf(self):
        print("Computing...")  # Only prints once
        return self.left is None and self.right is None

n = Node(10)
print(n.is_leaf)  # Prints "Computing..." then True
print(n.is_leaf)  # Returns True instantly from cache
```
