```python
# Arrays & Hashing

# Instead of O(N^2) nested loops, use a hash map for O(N) lookup.
# Tip: Use enumerate and dict comprehension logic on the fly

# 1. Two Sum (Hash Map Lookup)
def two_sum(nums, target):
    seen = {}  # val -> index
    for i, n in enumerate(nums):
        diff = target - n
        if diff in seen:
            return [seen[diff], i]
        seen[n] = i


nums, target = [2, 7, 11, 15], 9
two_sum(nums, target)
```

```python
# 3. Group Anagrams (Defaultdict + Sorted Tuple)
# Use collections.defaultdict to avoid key initialization checks.
# Use a sorted tuple as the key since lists are not hashable

from collections import defaultdict


def group_anagrams(strs):
    res = defaultdict(list)
    for s in strs:
        key = tuple(sorted(s))
        res[key].append(s)
    return list(res.values())


group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"])
```

```python
# 7. Encode / Decode Strings (Length Prefix)
# Use join with a demiliter that isn't in the charset, or encode length


def encode(strs):
    return "".join(f"{len(s)}#{s}" for s in strs)


def decode(s):
    res, i = [], 0
    while i < len(s):
        j = i
        while s[j] != "#":
            j += 1
        length = int(s[i:j])  # get s length for below slice
        res.append(s[j + 1 : j + 1 + length])
        i = j + 1 + length
    return res


print(encode(["lint", "code", "love", "you"]))
print(decode(encode(["lint", "code", "love", "you"])))
```

```python
# 8. Longest Consecutive Sequence (Set Lookups)
# Convert to a set for O(1) lookups.
# Only start counting if n-1 is not in the set (start of a sequence)


def longest_consecutive(nums):
    num_set = set(sorted(nums))  # sorted first to ensure ordering
    longest = 0
    for n in num_set:
        if n - 1 not in num_set:  # begin of new set
            length = 1
            while n + length in num_set:
                length += 1
            longest = max(longest, length)
    return longest


longest_consecutive([100, 4, 200, 1, 3, 2])
```

```python
# Two pointers

# 9. Valid Palindrome (Two Pointers)
# Use isalnum() and lower(). While loop is cleaner than for loop here.


def is_palindrome(s):
    l, r = 0, len(s) - 1
    while l < r:
        while l < r and not s[l].isalnum():
            l += 1
        while l < r and not s[r].isalnum():
            r -= 1
        if s[l].lower() != s[r].lower():
            return False
        l, r = l + 1, r - 1
    return True


is_palindrome("A man, a plan, a canal: Panama")
```

```python
# 10. Two sum II (Input Array Sorted)
# Classic left/right pointers converging.


def two_sum_sorted(nums, target):
    l, r = 0, len(nums) - 1
    while l < r:
        cur = nums[l] + nums[r]
        if cur > target:
            r -= 1
        elif cur < target:
            l += 1
        else:
            return [l + 1, r + 1]  # 1-indexed


two_sum_sorted([2, 7, 11, 15], 9)
```

```python
# 11. Three Sum (Sort + Two Pointes)
# Skip duplicates efficiently inside the loop


def three_sum(nums):
    res = []
    nums.sort()

    for i, a in enumerate(nums):
        if i > 0 and a == nums[i - 1]:
            continue  # Skip duplicates

        l, r = i + 1, len(nums) - 1
        while l < r:
            total = a + nums[l] + nums[r]
            if total > 0:
                r -= 1
            elif total < 0:
                l += 1
            else:
                res.append([a, nums[l], nums[r]])
                l += 1
                while nums[l] == nums[l - 1] and l < r:
                    l += 1

    return res


three_sum([-1, 0, 1, 2, -1, -4])
```

```python
# 12. Container With Most Water
# Move the pointer pointing to the shorter line.


def max_area(height):
    l, r = 0, len(height) - 1
    max_area = 0
    while l < r:
        area = (r - l) * min(height[l], height[r])
        max_area = max(max_area, area)
        if height[l] < height[r]:
            l += 1
        else:
            r -= 1
    return max_area


max_area([1, 8, 6, 2, 5, 4, 8, 3, 7])
```

```python
# 13. Trapping Rain Water
# No need for arrays, maintain max left and max right variables.


def trap(height):
    if not height:
        return 0

    l, r = 0, len(height) - 1
    left_max, right_max = height[l], height[r]
    res = 0

    while l < r:
        if left_max < right_max:
            l += 1
            left_max = max(left_max, height[l])
            res += left_max - height[l]
        else:
            r -= 1
            right_max = max(right_max, height[r])
            res += right_max - height[r]
    return res


trap([0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1])
```

```python
# Sliding Window

# 14. Best Time to Buy / Sell Stock
# A variation of sliding window. Track min price found so far.


def max_profit(prices):
    min_price = float("inf")
    max_profit = 0

    for p in prices:
        min_price = min(min_price, p)
        max_profit = max(max_profit, p - min_price)
    return max_profit


max_profit([7, 1, 5, 3, 6, 4])
```

```python
# 15. Longest Substring Without Repeating Characters
# Use a set and a sliding window, remove from left when duplicate found.


def length_of_longest_substring(s):
    char_set = set()
    l = 0
    res = 0

    for r in range(len(s)):
        while s[r] in char_set:
            char_set.remove(s[l])
            l += 1
        char_set.add(s[r])
        res = max(res, r - l + 1)
    return res


length_of_longest_substring("abcabcbb")
```

```python
# 16. Minimum Window Substring
# Use collections.Counter for need and a standard dict for have

from collections import Counter


def min_window(s, t):
    if len(t) > len(s):
        return ""

    count_t = Counter(t)
    need = len(count_t)
    window = {}
    have = 0

    res, res_len = [-1, -1], float("inf")
    l = 0

    for r, c in enumerate(s):
        window[c] = window.get(c, 0) + 1
        if c in count_t and window[c] == count_t[c]:
            have += 1
        while have == need:
            # Update result
            if (r - l + 1) < res_len:
                res = [l, r]
                res_len = r - l + 1
            # Pop left
            window[s[l]] -= 1
            if s[l] in count_t and window[s[l]] < count_t[s[l]]:
                have -= 1
            l += 1
    return s[res[0] : res[1] + 1] if res_len != float("inf") else ""


min_window("ADOBECODEBANC", "ABC")
```

```python
# Stack

# 17. Valid Parentheses
# Use a mapping dictionary map={')': '(', ...}, push open, pop close.


def is_valid_parentheses(s):
    stack = []
    mapping = {")": "(", "}": "{", "]": "["}
    for char in s:
        if char in mapping:  # Closing bracket
            top = stack.pop() if stack else "#"
            if mapping[char] != top:
                return False
        else:
            stack.append(char)
    return not stack


is_valid_parentheses("()[]{}")
```

```python
# 18. Min Stack
# Store tuples in the stack: (value, current_min)


class MinStack:
    def __init__(self):
        self.stack = []

    def push(self, val):
        if not self.stack:
            self.stack.append((val, val))
        else:
            current_min = min(val, self.stack[-1][1])
            self.stack.append((val, current_min))

    def pop(self):
        self.stack.pop()

    def top(self):
        return self.stack[-1][0]

    def getMin(self):
        return self.stack[-1][1]


m = MinStack()
m.push(-1)
m.push(0)
m.push(-3)
print(m.getMin())
```

```python
# 19. Evaluate Reverse Polish Notation
# use operator module or simple if/else with pop()


def eval_rpn(tokens):
    stack = []
    for t in tokens:
        if t not in "+-*/":
            stack.append(int(t))
        else:
            b, a = stack.pop(), stack.pop()
            if t == "+":
                stack.append(a + b)
            elif t == "-":
                stack.append(a - b)
            elif t == "*":
                stack.append(a * b)
            else:
                stack.append(int(a / b))  # truncates towards 0 with int()
    return stack[0]


eval_rpn(["2", "1", "+", "3", "*"])
```

```python
# 20. Daily Temperatures (Monotonic Stack)
# Stack stores indices, not values. Iterate in reverse or forward.
# Forward is standard: pop while current > stack pop


def daily_temperatures(temps):
    res = [0] * len(temps)
    stack = []  # stores indices

    for i, t in enumerate(temps):
        while stack and t > temps[stack[-1]]:
            prev_idx = stack.pop()
            res[prev_idx] = i - prev_idx
            print(i, t, res, stack)
        stack.append(i)
        print(">>", i, t, res, stack)

    return res


# stack stores indices, maintaining a decreasing temperature order
def daily_temperatures(temps):
    n = len(temps)
    ans = [0] * n
    stack = []  # monotonic stack
    for i in range(n):
        while stack and temps[i] > temps[stack[-1]]:  # new temp warmer, pop
            j = stack.pop()
            ans[j] = i - j
        stack.append(i)  # push current temp index
    return ans


def daily_temperatures_brute_force(temps):
    n = len(temps)
    ans = [0] * n
    for i in range(n):
        for j in range(i + 1, n):
            if temps[j] > temps[i]:
                ans[i] = j - i
                break
    return ans


print(daily_temperatures([73, 74, 75, 71, 69, 72, 76, 73]))
print(daily_temperatures_brute_force([73, 74, 75, 71, 69, 72, 76, 73]))
print(daily_temperatures_brute_force([73, 74, 75, 71, 69, 72, 76, 73]))
```

```python
# 21. Car Fleet
# Sort by position descending. If arrival time <= previous fleet time, they merge.


def car_fleet(target, position, speed):
    cars = sorted(zip(position, speed), reverse=True)
    stack = []
    for pos, sp in cars:
        time = (target - pos) / sp
        if not stack or time > stack[-1]:
            stack.append(time)
    return len(stack)


car_fleet(12, [10, 8, 0, 5, 3], [2, 4, 1, 1, 3])
```

```python
# Binary Search
# 22. Basic Binary Search
# while l <= r. Mid calculation l+ (r-1) // 2 prevents overflow (good style)


def binary_search(nums, target):
    l, r = 0, len(nums) - 1
    while l <= r:
        m = l + (r - l) // 2
        if nums[m] > target:
            r = m - 1
        elif nums[m] < target:
            l = m + 1
        else:
            return m
    return -1


binary_search([-1, 0, 3, 5, 9, 12], 9)
```

```python
# 23. Search in Rotated Sorted Array
# Check which half is sorted nums[l] <= nums[m],
# then check if target lies in that half


def search_rotated(nums, target):
    l, r = 0, len(nums) - 1

    while l <= r:
        m = (l + r) // 2
        if nums[m] == target:
            return m

        # Left sorted portion
        if nums[l] <= nums[m]:
            if target > nums[m] or target < nums[l]:
                l = m + 1
            else:
                r = m - 1
        # Right sorted portion
        else:
            if target < nums[m] or target > nums[r]:
                r = m - 1
            else:
                l = m + 1
    return -1


search_rotated([4, 5, 6, 7, 0, 1, 2], 0)
```

```python
# 24. Find Minimum in Rotated Array
# Compare nums[m] with nums[r]. If nums[m] > nums[r], min is to right.


def find_min(nums):
    l, r = 0, len(nums) - 1
    res = nums[0]
    while l <= r:
        if nums[l] < nums[r]:  # Sorted segment
            res = min(res, nums[l])
            break
        m = (l + r) // 2
        res = min(res, nums[m])
        if nums[m] >= nums[l]:
            l = m + 1
        else:
            r = m - 1
    return res


find_min([3, 4, 5, 1, 2])
```

```python
# 25. Linked List
# Reverse Linked List

# Iterate with prev, curr = curr, curr.next. Python allows one-line swappnig.


class ListNode:
    def __init__(self, val=0, next=None):
        self.val, self.next = val, next


def reverse_list(head):
    prev, curr = None, head

    while curr:
        nxt = curr.next
        curr.next = prev
        prev = curr
        curr = nxt
    return prev
```

```python
# 26. Merget Two Sorted Lists
# Use a dummy node to simplify edge cases


def merge_two_lists(l1, l2):
    dummy = ListNode()
    tail = dummy

    while l1 and l2:
        if l1.val < l2.val:
            tail.next = l1
            l1 = l1.next
        else:
            tail.next = l2
            l2 = l2.next
        tail = tail.next

    tail.next = l1 or l2  # Pythonic: pick the non-null one
    return dummy.next
```

```python
# 27. Reorder List (Fast/Slow Pointers)
# Find middle, reverse second half, merge.
# Use fast.next and fast.next.next check


def reorder_list(head):
    # 1. Find Middle
    slow, fast = head, head.next
    while fast and fast.next:
        slow, fast = slow.next, fast.next.next

    # 2. Reverse Second Half
    second = slow.next
    prev = slow.next = None  # Cut list
    while second:
        tmp = second.next
        second.next = prev
        prev = second
        second = tmp

    # 3. Merge
    first, second = head, prev
    while second:
        tmp1, tmp2 = first.next, second.next
        first.next = second
        second.next = tmp1
        first, second = tmp1, tmp2
```

```python
# 28. Remove Nth Node From End
# Dummy node + two pointers with a gap of n


def remove_nth_from_end(head, n):
    dummy = ListNode(0, head)
    left, right = dummy, head

    while n > 0:
        right = right.next
        n -= 1
    while right:
        left, right = left.next, right.next
    left.next = left.next.next
    return dummy.next
```

```python
# 29. Linked List Cycle (Floyd's Algo)
# slow moves 1, fast moves 2. If they meet, cycle exists.


def has_cycle(head):
    slow, fast = head, head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            return True
    return False
```

```python
# Trees

# 30. Binary Tree Level Order Traversal (BFS)
# Use collections.deque. Inner loop handles levels.

from collections import deque


def level_order(root):
    res = []
    q = deque([root]) if root else deque()

    while q:
        level = []
        for i in range(len(q)):
            node = q.popleft()
            level.append(node.val)
            if node.left:
                q.append(node.left)
            if node.right:
                q.append(node.right)
        res.append(level)
    return res
```

```python
# 31. Maximum Depth (DFS Recursive)
# One-liner recursion base case


def max_depth(root):
    if not root:
        return 0
    return 1 + max(max_depth(root.left), max_depth(root.right))
```

```python
# 32. Invert Binary Tree
# Swap children recursively or iterratively


def invert_tree(root):
    if not root:
        return None

    root.left, root.right = root.right, root.left  # Tuple swap
    invert_tree(root.left)
    invert_tree(root.right)
    return root
```

```python
# 33. Same Tree
# Return logic handles both base cases and value check concisely


def is_same_tree(p, q):
    if not p and not q:
        return True
    if not p or not q:
        return False
    if p.val != q.val:
        return False

    return is_same_tree(p.left, q.left) and is_same_tree(p.right, q.right)
```

```python
# 34. Subtree of Another Tree
# Combine is_same_tree logic with recursive traversal.


def is_subtree(root, sub_root):
    if not sub_root:
        return True
    if not root:
        return False

    if is_same_tree(root, sub_root):
        return True
    return is_subtree(root.left, sub_root) or is_subtree(root.right, sub_root)
```

```python
# 35. Lowest Common Ancestor (BST)
# Utilize BST property: if p, q > root, go right


def lowest_common_ancestor(root, p, q):
    cur = root
    while cur:
        if p.val > cur.val and q.val > cur.val:
            cur = cur.right
        elif p.val < cur.val and q.val < cur.val:
            cur = cur.left
        else:
            return cur
```

```python
# Heaps / Priority Queue

# 36. Kth Largest Element
# Use heapq which is a min-heap, keep heap size k.

import heapq


def find_kth_largest(nums, k):
    heap = []
    for n in nums:
        heapq.heappush(heap, n)
        if len(heap) > k:
            heapq.heappop(heap)  # Pop smallest
    return heap[0]  # largest
```

```python
# 37. Merge K Sorted Lists
# Push (val, list_node) into heap


def merge_k_lists(lists):
    dummy = ListNode()
    tail = dummy
    heap = []

    # Add heads of all lists to heap
    for i, node in enumerate(lists):
        if node:
            heapq.heappush(heap, (node.val, i, node))

    while heap:
        val, i, node = heapq.heappop(heap)
        tail.next = node
        tail = tail.next
        if node.next:
            heapq.heappush(heap, (node.next.val, i, node.next))
    return dummy.next
```

```python
# Graphs

# 38. Number of Islands (DFS)
# Modify grid in-place to mark visited ('1'->'0')


def num_islands(grid):
    if not grid:
        return 0
    rows, cols = len(grid), len(grid[0])
    count = 0

    def dfs(r, c):
        if r < 0 or c < 0 or r >= rows or c >= cols or grid[r][c] == "0":
            return
        grid[r][c] = "0"  # Sink island
        dfs(r + 1, c)
        dfs(r - 1, c)
        dfs(r, c + 1)
        dfs(r, c - 1)

    for r in range(rows):
        for c in range(cols):
            if grid[r][c] == "1":
                count += 1
                dfs(r, c)
    return count
```

```python
# 40. Pacific Atlantic Water Flow (Multi-source BFS/DFS)
# Run BFS/DFS from the oceans inwards.


def pacific_atlantic(heights):
    rows, cols = len(heights), len(heights[0])
    pac, atl = set(), set()

    def dfs(r, c, visited, prev_height):
        if (
            (r, c) in visited
            or r < 0
            or c < 0
            or r >= rows
            or c >= cols
            or heights[r][c] < prev_height
        ):
            return
        visited.add((r, c))
        dfs(r + 1, c, visited, heights[r][c])
        dfs(r - 1, c, visited, heights[r][c])
        dfs(r, c + 1, visited, heights[r][c])
        dfs(r, c - 1, visited, heights[r][c])

    for c in range(cols):
        dfs(0, c, pac, heights[0][c])  # Top row (Pacific)
        dfs(rows - 1, c, atl, heights[rows - 1][c])  # Bottom row (Atlantic)

    for r in range(rows):
        dfs(r, 0, pac, heights[r][0])  # Left col (Pacific)
        dfs(r, cols - 1, atl, heights[r][cols - 1])  # Right col (Atlantic)

    return list(pac.intersection(atl))
```

```python
# 41. Course Schedule (Topological Sort / Cycle Detection)
# Build adjacency list using defaultdict(set). Use a visited set or (state array)

from collections import defaultdict


def can_finish(num_courses, prerequisites):
    graph = defaultdict(list)
    for crs, pre in prerequisites:
        graph[crs].append(pre)

    visited = set()

    def dfs(crs):
        if crs in visited:
            return False  # Cycle
        if not graph[crs]:
            return True  # No prereqs

        visited.add(crs)
        for pre in graph[crs]:
            if not dfs(pre):
                return False
        visited.remove(crs)
        graph[crs] = []  # Optim: cache result
        return True

    for c in range(num_courses):
        if not dfs(c):
            return False
    return True


can_finish(2, [[1, 0]])
```

```python
# 42. Network Delay Time (Dijkstra)
# Use heapq , initialize distances with float('inf') .

import heapq
from collections import defaultdict


def network_delay_time(times, n, k):
    edges = defaultdict(list)
    for u, v, w in times:
        edges[u].append((v, w))

    min_heap = [(0, k)]  # (time, node)
    visit = set()
    t = 0

    while min_heap:
        w1, n1 = heapq.heappop(min_heap)
        if n1 in visit:
            continue
        visit.add(n1)
        t = w1  # Update total time

        for n2, w2 in edges[n1]:
            if n2 not in visit:
                heapq.heappush(min_heap, (w1 + w2, n2))

    return t if len(visit) == n else -1
```

```python
# 43. Union Find (Disjoint Set)
# Path compression and union by rank


class UnionFind:
    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n

    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])  # Path compression
        return self.parent[x]

    def union(self, x, y):
        root_x, root_y = self.find(x), self.find(y)
        if root_x == root_y:
            return False

        if self.rank[root_x] < self.rank[root_y]:
            self.parent[root_x] = root_y
        elif self.rank[root_x] > self.rank[root_y]:
            self.parent[root_y] = root_x
        else:
            self.parent[root_y] = root_x
            self.rank[root_x] += 1
        return True
```

```python
# Dynamic Programming

# 44. Climbing Stairs
# Space optimization (only need last 2 values)


def climb_stairs(n):
    one, two = 1, 1
    for _ in range(n - 1):
        one, two = one + two, one  # Tuple unpacking swap
    return one
```

```python
# 45. House Robber
# Similar to climbing stairs, maintain rob1 (max 2 steps back) and rob2 (max 1 step back)


def rob(nums):
    rob1, rob2 = 0, 0
    for n in nums:
        temp = max(rob1 + n, rob2)
        rob1 = rob2
        rob2 = temp
    return rob2
```

```python
# 46. Coin Change
# Bottom-up DP, initialize dp array with amount + 1 (a large dummy value)


def coin_change(coins, amount):
    dp = [amount + 1] * (amount + 1)
    dp[0] = 0

    for a in range(1, amount + 1):
        for c in coins:
            if a - c >= 0:
                dp[a] = min(dp[a], 1 + dp[a - c])

    return dp[amount] if dp[amount] != amount + 1 else -1
```

```python
# Backtracking

# 48. Subsets
# At each step, decide to incoude or exclude


def subsets(nums):
    res = []
    subset = []

    def dfs(i):
        if i >= len(nums):
            res.append(subset.copy())  # Copy is vital
            return

        # Include
        subset.append(nums[i])
        dfs(i + 1)

        # Exclude
        subset.pop()
        dfs(i + 1)

    dfs(0)
    return res
```
