```python
# Python mathematical operations

# 1. Basic Number Theory and Arithmetic

# 1. GCD
import math
from functools import reduce
from typing import Generator, List, Optional, Tuple

# Formula: gcd(a,b) = {a if b == 0 || gcd(b, a mod b) if b > 0}
# Step-by-Step Logic
# Divide $a$ by $b$ and find the remainder ($r$).
# # Replace $a$ with $b$.
# Replace $b$ with $r$.
# Repeat the process until $b$ becomes $0$.
# The value of $a$ at that moment is the GCD.


def gcd(a: int, b: int) -> int:
    # Compute the greatest common divisor using Euclid's algorithm.
    while b:
        a, b = b, a % b
    return a


# Test
assert gcd(48, 18) == 6
```

```python
# 2. LCM


def lcm(a: int, b: int) -> int:
    """Computes the least common multiple."""
    return abs(a * b) // gcd(a, b) if a and b else 0


assert lcm(4, 6) == 12
```

```python
# 3. Is Prime


def is_prime(n: int) -> bool:
    """Check primality efficiently using sqrt."""
    if n <= 1:
        return False
    if n <= 3:
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6
    return True


# Test
assert is_prime(17) is True
assert is_prime(18) is False
```

```python
# 4. primes_up_to


def prime_factors(n: int) -> List[int]:
    """Returns a list of prime factors of n."""
    factors = []
    d = 2
    temp = n
    while d * d <= temp:
        while temp % d == 0:
            factors.append(d)
            temp //= d
        d += 1
    if temp > 1:
        factors.append(temp)
    return factors


assert prime_factors(60) == [2, 2, 3, 5]
```

```python
# 5. Primes Up To


def primes_up_to(n: int) -> list[int]:
    """Sieve of Eratosthenes to find all primes up to n."""
    sieve = [True] * (n + 1)
    sieve[0] = sieve[1] = False
    for i in range(2, int(n**0.5) + 1):
        if sieve[i]:
            sieve[i * i : n + 1 : i] = [False] * len(sieve[i * i : n + 1 : i])
    return [i for i, prime in enumerate(sieve) if prime]


# Test
assert primes_up_to(10) == [2, 3, 5, 7]
```

```python
# 6. Factorial

from functools import lru_cache


@lru_cache
def factorial(n: int) -> int:
    """Recursive factorial with base case."""
    return 1 if n <= 1 else n * factorial(n - 1)


# Test
assert factorial(5) == 120
```

```python
# 7. Fibonacci


def fibonacci(n: int) -> int:
    """Returns the nth Fibonacci number iteratively (O(n))"""
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a


# Test
assert fibonacci(10) == 55
```

```python
# 8. Fibonacci Generator


def fibonacci_generator(n: int) -> Generator[int, None, None]:
    """Yields the first n Fibonacci number."""
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b


# Test
assert list(fibonacci_generator(5)) == [0, 1, 1, 2, 3]
```

```python
# 12. Armstrong Number


def is_armstrong(n: int) -> bool:
    """Checks if a number is an Armstrong number."""
    digits = [int(d) for d in str(n)]
    power = len(digits)
    return n == sum(d**power for d in digits)


# Test
assert is_armstrong(153) is True  # 1^3 + 5^3 + 3^3 = 153
```

```python
# 13. Is Perfect


def is_perfect(n: int) -> bool:
    """Checks if a number equals the sum of its proper divisors."""

    if n <= 1:
        return False
    return (
        sum(
            i
            for i in range(1, int(n**0.5) + 1)
            if n % i == 0
            for i in (i, n // i)
            if i != n
        )
        == n
    )
    # Corrected logic for proper divisors sum:
    proper_sum = sum(i for i in range(1, n) if n % i == 0)
    return proper_sum == n


# Test
assert is_perfect(28) is True
```

```python
# 2. Combinnatorics & Sequences

# 14. nCr


def nCr(n: int, r: int) -> int:
    """Calculates combinations count (n choose r)."""
    if r < 0 or r > n:
        return 0
    r = min(r, n - r)
    numer = reduce(lambda x, y: x * y, range(n, n - r, -1), 1)
    denom = math.factorial(r)
    return numer // denom


# Test
assert nCr(5, 2) == 10
```

```python
# 15. nPr


def nPr(n: int, r: int) -> int:
    """Calculate permutations count."""
    if r < 0 or r > n:
        return 0
    return reduce(lambda x, y: x * y, range(n, n - r, -1), 1)


# Test
assert nPr(5, 2) == 20
```

```python
# 16. Pascal Triangle

from typing import List


def pascals_triangle_row(n: int) -> List[int]:
    """Returns the nth row of Pascal's triangle (0-indexed)."""
    row = [1]
    for k in range(1, n + 1):
        # Calculate next element based on previous: row[k] = row[k-1] * (n-k+1)//k
        # Here using list comprehension update approach
        row = (
            [1] + [row[i] + row[i + 1] for i in range(len(row) - 1)] + [1]
            if k > 0
            else [1]
        )
        print(row)
    return row


# Test
assert pascals_triangle_row(4) == [1, 4, 6, 4, 1]
```

```python
# 20. Collatz Steps


def collatz_steps(n: int) -> int:
    """Returns number of steps to reach 1 in Collatz conjectures."""
    steps = 0
    while n != 1:
        n = 3 * n + 1 if n % 2 else n // 2
        steps += 1
    return steps


# Test
assert collatz_steps(27) == 111
```

```python
# 21. Arithmetic Progression


def arithmetic_progression(start: float, diff: float, n: int) -> List[float]:
    """Generates an arithmetic progression"""
    return [start + i * diff for i in range(n)]


# Test
assert arithmetic_progression(1, 2, 5) == [1, 3, 5, 7, 9]
arithmetic_progression(1, 3, 20)
```

```python
# 22. Geometric Progression


def geometric_progression(start: float, ratio: float, n: int) -> List[float]:
    """Generates a geometric progression"""
    return [start * (ratio**i) for i in range(n)]


# Test
assert geometric_progression(2, 3, 4) == [2, 6, 18, 54]
geometric_progression(1, 3, 20)
```

```python
# 3. Algebra & Geometry

# 24. Quadratic Roots
import cmath
from typing import Tuple


def quadratic_roots(a: float, b: float, c: float) -> Tuple[complex, complex]:
    """Solve ax^2 + bx + c = 0, returns complex roots."""
    discriminant = b**2 - 4 * a * c
    root1 = (-b + cmath.sqrt(discriminant)) / (2 * a)  # Needs cmath import
    root2 = (-b - cmath.sqrt(discriminant)) / (2 * a)
    return (root1, root2)


# Test
r1, r2 = quadratic_roots(1, -3, 2)  # Roots 2, 1
assert abs(r1 - 2) < 1e-9 or abs(r2 - 2) < 1e-9
```

Euclidean Distance Formula

The Formula (for 3D):$$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2 + (z_2 - z_1)^2}$$

```python
# 25. Euclidean Distance


def euclidean_distance(p1: Tuple[float, ...], p2: Tuple[float, ...]) -> float:
    """Calculate Euclidean distance between two points"""
    return math.sqrt(sum((a - b) ** 2 for a, b in zip(p1, p2)))


# Test
assert euclidean_distance((0, 0), (3, 4)) == 5
```

```python
# 27. Triangle Area Heron


def triangle_area_heron(a: float, b: float, c: float) -> float:
    """Calculate area of triangle given side lengths
    using Heron's formula"""
    s = (a + b + c) / 2
    area = math.sqrt(s * (s - a) * (s - b) * (s - c))
    return area


# Test
assert abs(triangle_area_heron(3, 4, 5) - 6) < 1e-9
```

```python
# 33. Dot Product


def dot_product(v1: List[float], v2: List[float]) -> float:
    """Compute dot product"""
    return sum(a * b for a, b in zip(v1, v2))


# Test
assert dot_product([1, 2], [3, 4]) == 11
dot_product([1, 1], [2, 2])
```

```python
# 34. Cross Product 3D


def cross_product_3d(v1: List[float], v2: List[float]) -> List[float]:
    """Computes cross products for 3Dvectors."""
    x = v1[1] * v2[2] - v1[2] * v2[1]
    y = v1[2] * v2[0] - v1[0] * v2[2]
    z = v1[0] * v2[1] - v1[1] * v2[0]
    return [x, y, z]


# Test
assert cross_product_3d([1, 0, 0], [0, 1, 0]) == [0, 0, 1]
cross_product_3d([1, 0, 0], [0, 1, 0])
```

```python
# 35. Magnitude


def magnitude(v: List[float]) -> float:
    """Returns the magnitude (norm) of a vector."""
    return math.sqrt(sum(x**2 for x in v))


# Test
assert magnitude([3, 4]) == 5
magnitude([3, 4])
```

```python
# 36. Normalize


def normalize(v: List[float]) -> List[float]:
    """Normalizes a vector to unit length."""
    mag = magnitude(v)
    if mag == 0:
        return v
    return [x / mag for x in v]


# Test
assert magnitude(normalize([3, 4])) == 1.0
normalize([1, 1])
```

```python
# 37. Matrix Transpose


def matrix_transpose(matrix: List[List[float]]) -> List[List[float]]:
    """Transposes a matrix using zip"""
    return [list(row) for row in zip(*matrix)]


# Test
assert matrix_transpose([[1, 2], [3, 4]]) == [[1, 3], [2, 4]]
matrix_transpose([[1, 2], [3, 4]])
```

```python
# 38. Matrix Multiply


def matrix_multiply(m1: List[List[float]], m2: List[List[float]]) -> List[List[float]]:
    """Multiplies two matrices"""
    m2_t = matrix_transpose(m2)
    return [[dot_product(row, col) for col in m2_t] for row in m1]


# Test
# 2x3 * 2x2 = 2x2
res = matrix_multiply([[1, 2, 3], [4, 5, 6]], [[7, 8], [9, 1], [2, 3]])
assert res == [[31, 19], [85, 55]]
matrix_multiply([[1, 2, 3], [4, 5, 6]], [[7, 8], [9, 1], [2, 3]])
```

```python
# 40. Identity Matrix


def identity_matrix(n: int) -> List[List[int]]:
    """Creates an NxN identity matrix."""
    return [[1 if i == j else 0 for j in range(n)] for i in range(n)]


# Test
assert identity_matrix(2) == [[1, 0], [0, 1]]
identity_matrix(2)
```

```python
# 5. Statistics & Probability

# 41. Mean
import statistics


def calculate_mean(data: List[float]) -> float:
    """Calculate arithmetic mean."""
    return sum(data) / len(data) if data else 0


# Test
assert calculate_mean([1, 2, 3, 4]) == 2.5
```

```python
# 42. Median


def calculate_median(data: List[float]) -> float:
    """Calculate median without importing statistics module."""
    sorted_data = sorted(data)
    n = len(sorted_data)
    mid = n // 2
    if n % 2 == 0:
        return (sorted_data[mid - 1] + sorted_data[mid]) / 2
    return sorted_data[mid]


# Test
assert calculate_median([1, 3, 2, 4]) == 2.5
assert calculate_median([1, 2, 3]) == 2
calculate_median([1, 2, 3, 4])
```

```python
# 43. Mode

from typing import Optional


def calculate_mode(data: List[float]) -> Optional[float]:
    """Find the most frequent element."""
    if not data:
        return None
    counts = {}
    for x in data:
        counts[x] = counts.get(x, 0) + 1
    return max(counts, key=counts.get)


# Test
assert calculate_mode([1, 2, 2, 3]) == 2
calculate_mode([1, 2, 2, 3])
```

```python
# 44. Variance


def variance(data: List[float], population: bool = True) -> float:
    """Calculates variance."""
    m = calculate_mean(data)
    n = len(data)
    denom = n if population else n - 1
    return sum((x - m) ** 2 for x in data) / denom


# Test
pop_var = variance([1, 2, 3, 4])
assert abs(pop_var - 1.25) < 1e-9
```

```python
# 45. Std_dev


def std_dev(data: List[float], population: bool = True) -> float:
    """Calculate standard deviation."""
    return math.sqrt(variance(data, population))


# Test
assert abs(std_dev([1, 2, 3, 4]) - math.sqrt(1.25)) < 1e-9
std_dev([1, 2, 3, 4])
```

```python
# 46. Z scores


def z_scores(data: List[float]) -> List[float]:
    """Calculates z-scores for a dataset."""
    m = calculate_mean(data)
    s = std_dev(data)
    return [(x - m) / s for x in data]


z = z_scores([1, 2, 3, 4])
print(z)
assert abs(z[0] + 1.341640786) < 1e-5  # Approx check
```

```python
# 47. Covariance


def covariance(x: List[float], y: List[float]) -> float:
    """Calculates sample covariance"""
    n = len(x)
    mean_x, mean_y = calculate_mean(x), calculate_mean(y)
    return sum((xi - mean_x) * (yi - mean_y) for xi, yi in zip(x, y)) / (n - 1)


# Test
cov = covariance([1, 2, 3], [4, 5, 6])
print(cov)
assert abs(cov - 1.0) < 1e-9
```

```python
# 48. Correlation


def correlation(x: List[float], y: List[float]) -> float:
    """Calculates Pearson correlation coefficient."""
    cov = covariance(x, y)
    std_x, std_y = std_dev(x, population=False), std_dev(y, population=False)
    return cov / (std_x * std_y)


# Test
corr = correlation([1, 2, 3], [4, 5, 6])
print(corr)
assert abs(corr - 1.0) < 1e-9
```

```python
# 6. Calculus & Numerical Methods

# 49. Derivative

from typing import Callable


def derivative(f: Callable[[float], float], x: float, h: float = 1e-6) -> float:
    """Numerical derivative using central difference."""
    return (f(x + h) - f(x - h)) / (2 * h)


# Test
assert abs(derivative(lambda x: x**2, 2) - 4) < 1e-5
```

```python
# 50. Integral Trapz


def integral_trapz(
    f: Callable[[float], float], a: float, b: float, n: int = 1000
) -> float:
    """Numerical integration using Trapezoidal rule."""
    step = (b - a) / n
    area = 0.5 * (f(a) + f(b))
    for i in range(1, n):
        area += f(a + i * step)
    return area * step


# Test
integral_val = integral_trapz(lambda x: x**2, 0, 2)
# Exact integral is 2^3/3= 8/3 = 2.666...
assert abs(integral_val - 2.666) < 0.01
```

```python
# 51. Newton Raphson


def newton_raphson(
    f: Callable[[float], float],
    df: Callable[[float], float],
    x0: float,
    tol: float = 1e-6,
) -> float:
    """Find root of f near x0."""
    x = x0
    while abs(f(x)) > tol:
        x = x - f(x) / df(x)
    return x


# Test
root = newton_raphson(lambda x: x**2 - 2, lambda x: 2 * x, 1)
print(root)
assert abs(root - math.sqrt(2)) < 1e-6
```

```python
# 52. Taylor Series Approx


def taylor_series_approx(
    f: Callable[[float], float],
    derivatives: List[Callable[[float], float]],
    a: float,
    x: float,
) -> float:
    """Calculates Taylor series approximation at x centered at a."""
    result = f(a)
    for n, deriv in enumerate(derivatives, 1):
        result += deriv(a) * (x - a) ** n / math.factorial(n)
    return result


# Test: ^x at a=0. derivatives are all e^x (lambda x: 1, lambda x: 1, etc)
approx = taylor_series_approx(lambda x: 1, [lambda x: 1, lambda x: 1], 0, 0.5)
print(approx)
assert abs(approx - 1.625) < 1e-9  # 1 + x + x^2/3 = 1 + 0.5 + 0.125
```

```python
# 55. Monotonic


def is_monotonic(seq: List[float]) -> bool:
    """Checks if sequence is monotonic (increasing or descreasing)."""
    return all(seq[i] <= seq[i + 1] for i in range(len(seq) - 1)) or all(
        seq[i] >= seq[i + 1] for i in range(len(seq) - 1)
    )


# Test
assert is_monotonic([1, 2, 3, 3]) is True
assert is_monotonic([3, 2, 1]) is True
assert is_monotonic([1, 3, 2]) is False
```

```python
# 57. Moving Average


def moving_average(data: List[float], window: int) -> List[float]:
    """Calculate simple moving average."""
    ret = []
    for i in range(len(data) - window + 1):
        ret.append(sum(data[i : i + window]) / window)
    return ret


# Test
assert moving_average([1, 2, 3, 4, 5], 3) == [2.0, 3.0, 4.0]
```

```python
# 59. Relu


def relu(x: float) -> float:
    """Relu activation function."""
    return max(0, x)


# Test
assert relu(-5) == 0
assert relu(5) == 5
```
